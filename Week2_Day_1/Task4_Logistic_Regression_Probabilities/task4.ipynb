```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer, load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
```

```python
print("\n===== BREAST CANCER DATASET =====")
data = load_breast_cancer()

X = data.data
y = data.target

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)
binary_model = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=10000)
)

binary_model.fit(X_train, y_train)
predictions = binary_model.predict(X_test)
accuracy = accuracy_score(y_test, predictions)

print("Dataset Shape:", X.shape)
print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
print("Accuracy:", round(accuracy, 4))
```

```python
print("\n===== FIRST 5 TEST ROW PROBABILITIES =====")
probabilities = binary_model.predict_proba(X_test[:5])
probability_table = pd.DataFrame(
    probabilities,
    columns=[
        f"Probability: {name}"
        for name in data.target_names
    ]
)

probability_table["Predicted Class"] = binary_model.predict(X_test[:5])
probability_table["Actual Class"] = y_test[:5]
print(probability_table.to_string(index=False))
```

```python
print("\n===== THRESHOLD COMPARISON =====")
class_1_probabilities = binary_model.predict_proba(X_test)[:, 1]

predictions_05 = (class_1_probabilities >= 0.5).astype(int)
predictions_03 = (class_1_probabilities >= 0.3).astype(int)

threshold_table = pd.DataFrame({
    "Actual": y_test[:10],
    "Prediction (0.5)": predictions_05[:10],
    "Prediction (0.3)": predictions_03[:10],
    "Probability of Class 1": class_1_probabilities[:10]
})
print(threshold_table.to_string(index=False))
changed = np.sum(predictions_05 != predictions_03)
print("\nNumber of changed predictions:", changed)
```

```python
print("\n===== IRIS MULTI-CLASS DATASET =====")
iris = load_iris()

X_iris = iris.data
y_iris = iris.target

X_train_iris, X_test_iris, y_train_iris, y_test_iris = train_test_split(
    X_iris,
    y_iris,
    test_size=0.20,
    random_state=42,
    stratify=y_iris
)

iris_model = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=10000)
)

iris_model.fit(X_train_iris, y_train_iris)
iris_probabilities = iris_model.predict_proba(X_test_iris[:3])

iris_table = pd.DataFrame(
    iris_probabilities,
    columns=iris.target_names
)

iris_table.insert(
    0,
    "Flower",
    ["Flower 1", "Flower 2", "Flower 3"]
)

iris_table["Predicted Class"] = iris_model.predict(X_test_iris[:3])
print("\nProbability Table for 3 Flowers:")
print(iris_table.to_string(index=False))
```

```python
print("\n===== WHY PROBABILITIES ARE USEFUL =====")

print("1. Probabilities show how confident a model is in its prediction.")
print("2. They help us change the decision threshold based on the problem.")
print("3. They help compare uncertainty between different predictions.")
```
